```python
# BMI Calculator
```

```python
name = input("Enter your name: ")
weight = int(input("Enter weight in lbs: "))
height = int(input("Enter height in inches: "))
BMI = round((weight * 703) / (height ** 2),1)

if BMI < 18.5:
    classification = "you are underweight."
elif BMI <= 24.9:
    classification = "you have a normal weight."
elif BMI <= 29.9:
    classification = "you are overweight."
elif BMI <= 34.9:
    classification = "you are obese."
elif BMI <= 39.9:
    classification ="you are severely obese."
else:
    classification = "you are morbidly obese."

print(f"{name}, your BMI is {BMI}. This means {classification}")
```
